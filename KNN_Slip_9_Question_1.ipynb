```python
data['target']=iris.target
data0=data[:50]
data1=data[50:100]
```

```python
import matplotlib.pyplot  as plt

plt.xlabel('sepal length')
plt.ylabel('sepal width')

plt.scatter(data0['sepal length (cm)'] , data0['sepal width (cm)'],color="red",marker="*")
plt.scatter(data1['sepal length (cm)'] , data1['sepal width (cm)'] , color="orange" , marker="+")
```

```python
plt.xlabel('petal length')
plt.ylabel('petal width')

plt.scatter(data0['petal length (cm)'] , data0['petal width (cm)'],color="red",marker="*")
plt.scatter(data1['petal length (cm)'],data1['petal width (cm)'],color="green",marker=".")
```

```python
from sklearn.model_selection import train_test_split
x = iris.data
y = iris.target
x_train ,x_test , y_train , y_test = train_test_split(x, y , test_size=0.2)
x_train
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=10)
knn.fit(x_train , y_train)
```

```python
knn.score(x_test , y_test)
```

```python
knn.predict([[4.8 , 3.8 , 1.5 , 0.3]])
```

```python
from sklearn.metrics import confusion_matrix
y_pred = knn.predict(x_test)
cm = confusion_matrix(y_test , y_pred)
cm
```

```python
import matplotlib.pyplot  as plt

import seaborn as sn
plt.figure(figsize=(7 , 5))
sn.heatmap(cm , annot=True)
plt.xlabel('predicted')
plt.ylabel('truth')
```
